```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("HCL.csv")
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='%d-%b-%y')
df = df.sort_values('Date')
```

```python
plt.figure()
plt.plot(df['Date'], df['Close Price'])
plt.title("HCL Stock Closing Price Trend")
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure()
plt.scatter(df['Open Price'], df['Close Price'])
plt.title("Open Price vs Close Price")
plt.xlabel("Open Price")
plt.ylabel("Close Price")
plt.show()
```

```python
plt.figure()
plt.hist(df['Close Price'], bins=20)
plt.title("Distribution of Close Price")
plt.xlabel("Close Price")
plt.ylabel("Frequency")
plt.show()
```

```python
top10 = df.head(10)

plt.figure()
plt.bar(top10['Date'].astype(str), top10['Total Traded Quantity'])
plt.title("Top 10 Days - Traded Quantity")
plt.xticks(rotation=45)
plt.show()
```

```python
top5 = df.head(5)

plt.figure()
plt.pie(top5['Total Traded Quantity'], labels=top5['Date'].astype(str), autopct='%1.1f%%')
plt.title("Trading Volume Share")
plt.show()
```

```python
plt.figure()
plt.boxplot(df['Close Price'])
plt.title("Box Plot of Close Price")
plt.show()
```

```python
plt.figure()
plt.plot(df['Date'], df['High Price'], label='High Price')
plt.plot(df['Date'], df['Low Price'], label='Low Price')
plt.legend()
plt.title("High vs Low Price")
plt.xticks(rotation=45)
plt.show()
```
